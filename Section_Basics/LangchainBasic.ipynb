### Interactive with LLM

```python
from langchain_ollama import ChatOllama

llm = ChatOllama(
base_url="http://localhost:11434",
model="qwen2:7b",
temperature=0.5,
num_predict=512
)

response=llm.invoke("Write a poem about a lonely computer in the style of Shakespeare.")
print(response)
```

### Load ENV file

```python
from dotenv import find_dotenv, load_dotenv
import os

dotenv_path = find_dotenv(usecwd=True)
if not dotenv_path:
    raise RuntimeError("Could not find .env")

load_dotenv(dotenv_path, override=True)

assert os.getenv("LANGSMITH_TRACING", "").lower() == "true"
assert os.getenv("LANGSMITH_API_KEY"), "LANGSMITH_API_KEY is missing"
```

### Prompt & Chat Template


```python
from langchain_core.prompts import PromptTemplate
prompt = PromptTemplate.from_template("Write a poem about a lonely computer in the style of Shakespeare.")
response = llm.invoke(prompt.format())
print(response)
```
